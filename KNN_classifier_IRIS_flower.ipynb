**4.Data Preprocessing**

```python
from sklearn import preprocessing
le=preprocessing.LabelEncoder()
le.fit(df.species)
df['species']=le.transform(df.species)
print(df)
```

```python
print(df.species)
```

**Separate features and target variable**

```python
X=df.iloc[:,0:4]
```

```python
X
```

```python
y=df.iloc[:,4]
```

```python
y
```

**5.Split the data into training and testing**

```python
x_train,x_test,y_train,y_test=train_test_split(X,y,train_size=0.70,random_state=42)
```

```python
print(x_train.shape)
```

```python
print(x_test.shape)
```

```python
print(y_train.shape)
```

```python
print(y_test.shape)
```

```python
y_train
```

**6.Building the model-k-Nearest Neighbors algorithm**

```python
model=KNeighborsClassifier(n_neighbors=5)
```

```python
model
```

**7.Train the model with training data**

```python
model_train=model.fit(x_train,y_train)
```

```python
model_train
```

**bold text**

```python
y_pred=model_train.predict(x_test)
```

```python
y_pred
```

```python
y_pred[0:10]
```

```python
i=0
print("\n----------------------------------------------------------------------------------------------")
print('%-25s %-25s %25s' % ('Orginial Label','Predicted Label','Correct/Wrong'))
print("------------------------------------------------------------------------------------------------")
for label in y_test:
  print('%-25s %-25s' % (label,y_pred[i]), end="")
  if(label==y_pred[i]):
    print('%-25s' % ('Correct'))
  else:
    print('%-25s' %('Wrong'))
  i=i+1
print("\n----------------------------------------------------------------------------------------------")
```

```python
print(y_test[:10])
```

```python
import sklearn.metrics as metrics
confusion=metrics.confusion_matrix(y_test,y_pred)
confusion
```

```python
accuracy=accuracy_score(y_test,y_pred)
```

```python
accuracy
```

```python
print('Accuracy of the classifier is  %0.2f' % metrics.accuracy_score(y_test,y_pred))
print("-------------------------------------------------------------------------------------------")
```

```python
print("\nClassification Report:\n",)
class_wise=metrics.classification_report(y_test,y_pred)
print(class_wise)
```
